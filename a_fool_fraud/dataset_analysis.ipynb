# Dataset analysis - Credit Card Fraud
This notebook performs EDA on the credit card fraud dataset.

**Dataset URL:** https://storage.googleapis.com/qwasar-public/track-ds/creditcard.csv

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

url = 'https://storage.googleapis.com/qwasar-public/track-ds/creditcard.csv'
df = pd.read_csv(url)
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
# Class distribution
print(df['Class'].value_counts())
print('\nPercentages:')
print(df['Class'].value_counts(normalize=True) * 100)
```

```python
# Basic statistics
print(df.describe())
```

```python
# Correlation heatmap (sampled for performance)
sample = df.sample(n=20000, random_state=42)  # sample to speed up heatmap
plt.figure(figsize=(12,8))
sns.heatmap(sample.corr(), cmap='viridis')
plt.title('Correlation matrix (sample)')
```

```python
# Amount distribution
plt.figure(figsize=(8,4))
sns.histplot(df['Amount'], bins=50)
plt.title('Transaction Amount distribution')
```

```python
# Time distribution
plt.figure(figsize=(8,4))
sns.histplot(df['Time'], bins=50)
plt.title('Transaction Time distribution')
```

```python
# Focused cleaning: nulls, duplicates, basic checks
import pandas as pd

print("Initial shape:", df.shape)

# Missing values
print("\nMissing values per column:")
print(df.isnull().sum())

# Duplicates
print("\nAny duplicates?", df.duplicated().any())
if df.duplicated().any():
    ndup = df.duplicated().sum()
    print("Number of duplicate rows:", ndup)
    df = df.drop_duplicates().reset_index(drop=True)
    print("Shape after dropping duplicates:", df.shape)
else:
    print("No duplicates found")

# Basic dtype checks & sample
print("\nData types:")
print(df.dtypes)
print("\nSample rows:")
display(df.head(3))

# Minor value hygiene examples (no-op if not required)
# Ensure Class is integer
df['Class'] = df['Class'].astype(int)

# Save cleaned local copy for downstream notebooks
out_path = "./creditcard_full.csv"
df.to_csv(out_path, index=False)
print(f"\nSaved cleaned file to: {out_path}")
```
