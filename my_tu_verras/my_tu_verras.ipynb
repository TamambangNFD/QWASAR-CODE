```python
# Clean dataset
def clean_dataset(boston_dataframe):
    columns_to_clean = boston_dataframe.columns 
    for col in columns_to_clean:
        median_val = boston_dataframe[col].median()
        boston_dataframe[col].fillna(median_val, inplace=True)
    return boston_dataframe
cleaned_df = clean_dataset(boston_dataframe)
print(cleaned_df.head(10))
```

```python
#Plot Histograms

def print_histograms(boston_dataframe):
    columns = boston_dataframe.columns
    n_cols = 4
    n_rows = (len(columns) + n_cols - 1) // n_cols
    
    plt.figure(figsize=(16, 4 * n_rows))
    
    for i, col in enumerate(columns):
        plt.subplot(n_rows, n_cols, i + 1)
        plt.hist(boston_dataframe[col].dropna(), bins=20, color='skyblue', edgecolor='black')
        plt.title(f'Histogram of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
    
    plt.tight_layout()
    plt.show()
```

```python
# Compute correlations matrix

def compute_correlations_matrix(boston_dataframe):
    # Return correlation matrix as a DataFrame (so .shape works in tests)
    return boston_dataframe.corr(method='pearson')


def print_scatter_matrix(boston_dataframe):
    """
    Plot scatter matrix for all numeric columns in the dataset.
    Histograms appear on the diagonal.
    """
    numeric_df = boston_dataframe.select_dtypes(include=['number'])
    if numeric_df.shape[1] < 2:
        print("Need at least two numeric columns to plot scatter matrix.")
        return

    scatter_matrix(numeric_df, figsize=(12, 12), diagonal='hist', hist_kwds={"bins": 20})
    plt.tight_layout()
    plt.show()


# ---- Main execution ----
boston_dataframe = load_dataset()
print("Columns in dataset:", boston_dataframe.columns.tolist())  # 👈 debug step

cleaned_df = clean_dataset(boston_dataframe)

# Compute correlations on cleaned data
correlations = compute_correlations_matrix(cleaned_df)

```

```python
def boston_fit_model(boston_dataframe):
  # SELECT two columns from our
  model_dataset = boston_dataframe[["RM","MDEV"]]
  regressor = linear_model.LinearRegression()
  # Extract column 1
  x = model_dataset.iloc[:, :-1].values  # all rows, all columns except last → "RM"
  # Extract column 2
  y = model_dataset.iloc[:, 1].values  # all rows, second column → "MDEV"
  # Train the model
  regressor.fit(x, y)
  return regressor
```

```python
# predict
def boston_predict(estimator, array_to_predict):
    new_array = np.array(array_to_predict).reshape(1, -1) 
    prediction = estimator.predict(new_array)
    return prediction
```
